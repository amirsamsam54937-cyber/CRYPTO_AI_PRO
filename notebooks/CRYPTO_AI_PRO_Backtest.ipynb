```python
# تنظیمات بک‌تست — قبل از اجرا تغییر بده
SYMBOL = "BTCUSDT"       # مثال: ETHUSDT, SOLUSDT
DAYS = 90                 # بازه تاریخی
LEVERAGE = 3              # سقف اهرم شبیه‌سازی
RISK_PER_TRADE = 0.01     # ۱ درصد سرمایه
TAKER_FEE = 0.0004        # کارمزد هر سمت؛ مطابق کارمزد واقعی خودت تنظیم کن
SLIPPAGE_BPS = 2          # لغزش فرضی بر حسب basis points
MAX_HOLD_BARS = 96        # ۹۶ کندل ۱۵دقیقه‌ای = ۲۴ ساعت
OUTPUT = "backtest_trades.csv"
```

```python
!python tools/backtest.py --symbol {SYMBOL} --days {DAYS} --leverage {LEVERAGE} --risk {RISK_PER_TRADE} --fee {TAKER_FEE} --slippage-bps {SLIPPAGE_BPS} --max-hold-bars {MAX_HOLD_BARS} --out {OUTPUT}
```

```python
import pandas as pd
trades = pd.read_csv(OUTPUT)
display(trades.tail(20))
print("تعداد معاملات:", len(trades))
if len(trades):
    print("میانگین R خالص:", round(trades.net_r.mean(), 3))
    print("درصد برد:", round((trades.net_r > 0).mean() * 100, 2))
```
